```python
def sum_proper(n):
    """Calculate sum of proper divisors of n (s(n) = sigma(n)-n)"""
    if n <= 1:
        return 0
    total = 0
    for d in range(1, n):
        if n % d == 0:
            total += d
    return total

test_numbers = [6,8,220,284,7]
for num in test_numbers:
    s = sum_proper(num)
    print(f"s({num}) = {s}")
```

```python
# Define a function to detect loops in an aliquot sequence
# n is the starting number and k is the maximum number of terms
def detect_loop(n, k=100000):

    # Use the function from Number 2 to generate the aliquot sequence
    sequence = aliquot_sequence(n, k)

    # Create an empty dictionary to store numbers we have already seen
    seen = {}

    # Go through each number in the sequence, keeping track of its position
    for i, num in enumerate(sequence):

        # Check whether the current number has appeared before
        if num in seen:
            cycle = sequence[seen[num]:i]

            # Return the results because a loop has been detected
            return {
                "status": "Loop detected",
                "sequence": sequence[:i],
                "cycle": cycle
            }

        # Store the number and its position
        seen[num] = i

    return {
        "status": "No loop detected within the computed terms",
        "sequence": sequence,
        "cycle": None
    }
```
