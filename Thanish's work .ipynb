```python
def sum_proper(n):
    """Calculate sum of proper divisors of n (s(n) = sigma(n)-n)"""
    if n <= 1:
        return 0
    total = 0
    for d in range(1, n):
        if n % d == 0:
            total += d
    return total

test_numbers = [6,8,220,284,7]
for num in test_numbers:
    s = sum_proper(num)
    print(f"s({num}) = {s}")
```
