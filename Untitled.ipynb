# Part 1: Project Background and Mathematical Definitions

## Introduction
This group project investigates aliquot sequences. An aliquot sequence is constructed by repeatedly calculating the sum of proper divisors of an integer. We will write code to generate aliquot sequences, classify their behaviour, study cycles such as amicable pairs, and visualise our results for a range of starting integers.

All computations follow fixed stopping rules:
- Maximum number of terms in one sequence: 30
- Upper numerical threshold: if any term exceeds $10^9$, the sequence calculation terminates early.

## Key Mathematical Definitions
1. Sum-of-divisors function $\sigma(n)$
The divisor-sum function $\sigma(n)$ returns the sum of all positive divisors of integer $n$, including $n$ itself.
$$\sigma(n)=\sum_{d\mid n} d$$

2. Sum of proper divisors $s(n)$
Proper divisors are positive integers less than $n$ which divide $n$.
$$s(n)=\sigma(n)-n$$
Note that $n=1$ has no proper divisors, so $s(1)=0$.

3. Aliquot sequence
Given a starting positive integer $k$, an aliquot sequence is defined iteratively:
$$
\begin{cases}
a_0 = k \\
a_{i+1}=s(a_i)
\end{cases}
$$
We repeatedly apply the function $s$ to generate successive terms.

4. Three possible behaviours of aliquot sequences
Within our stopping constraints, every aliquot sequence has one of three outcomes:
1. **Terminate**: The sequence eventually reaches 0. For example: $8 \to 7 \to 1 \to 0$.
2. **Enter a cycle**: Values start repeating. Once a value repeats, a cycle is detected.
    - Cycle length 1: perfect numbers: $s(n)=n$
    - Cycle length 2: amicable pairs e.g. $220 \leftrightarrow 284$
    - Cycle length $\ge 3$: sociable numbers.
3. **Unbounded growth**: Sequence terms keep increasing and exceed $10^9$. We stop before observing termination or a cycle.

If none of these conditions are met after 30 terms, we also stop the computation.

5. Perfect, abundant and deficient numbers
We classify positive integers using $s(n)$:
- Perfect number: $s(n)=n$ (sum of proper divisors equals itself)
- Abundant number: $s(n) > n$ (sum of proper divisors is larger)
- Deficient number: $s(n) < n$ (sum of proper divisors is smaller)

## Task
Write a Python function `sum_proper(n)` to compute $s(n)$. We test it on standard known examples:
- Perfect number: $s(6)=1+2+3=6$
- Amicable pair: $s(220)=284,\ s(284)=220$
- Prime number: $s(p)=1$ (the only proper divisor of a prime is 1)# Part 1: Project Background and Mathematical Definitions


```python
def sum_proper(n):
    """Calculate sum of proper divisors of n (s(n) = sigma(n)-n)"""
    if n <= 1:
        return 0
    total = 0
    for d in range(1, n):
        if n % d == 0:
            total += d
    return total

test_numbers = [6,8,220,284,7]
for num in test_numbers:
    s = sum_proper(num)
    print(f"s({num}) = {s}")
```

# Part 2: Computing Aliquot Sequences

## 2.1 Introduction

An aliquot sequence is generated by repeatedly applying the sum of proper divisors function to the previous term.

Given a starting positive integer n, the next term of the sequence is calculated using the function s(n), which was implemented in Part 1.

The process continues by repeatedly applying this function to generate successive terms.

In this section, we develop a Python function to compute aliquot sequences subject to two computational constraints:

- The maximum number of terms is limited to k = 30.
- The computation stops when a term exceeds the threshold i = 10^9.

The sequence also stops naturally when it reaches zero.

The function developed in this section will be used in subsequent parts of the project to investigate and classify aliquot sequences.

## 2.2 Mathematical Definition

Let the starting positive integer be n. An aliquot sequence is defined recursively as:

$$
a_0 = n
$$

$$
a_{j+1} = s(a_j), \quad j \geq 0
$$

where s(a_j) represents the sum of all proper divisors of the previous term.

For example, consider an aliquot sequence starting at n = 12:

$$
a_0 = 12
$$

$$
a_1 = s(12) = 1+2+3+4+6 = 16
$$

$$
a_2 = s(16) = 1+2+4+8 = 15
$$

$$
a_3 = s(15) = 1+3+5 = 9
$$

Continuing this process gives:

$$
12 \rightarrow 16 \rightarrow 15 \rightarrow 9
\rightarrow 4 \rightarrow 3 \rightarrow 1 \rightarrow 0
$$

The sequence terminates when it reaches zero.

In our implementation, the computation is also restricted by a maximum number of terms k = 30 and an upper threshold i = 10^9.

```python
def aliquot_sequence(n, k=30, i=10**9):
    """
    Generate an aliquot sequence starting at n.

    Parameters:
    n : starting positive integer
    k : maximum number of terms
    i : upper numerical threshold

    Returns:
    A list containing the aliquot sequence.
    """

    if n < 1:
        raise ValueError("n must be a positive integer")

    sequence = [n]

    while len(sequence) < k:

        current = sequence[-1]

        # Stop if the current term is zero or exceeds the threshold
        if current == 0 or current > i:
            break

        # Calculate the next term using Part 1
        next_term = sum_proper(current)

        sequence.append(next_term)

        # Stop if the new term is zero or exceeds the threshold
        if next_term == 0 or next_term > i:
            break

    return sequence
```

```python
test_numbers = [12, 7, 6, 1]

for n in test_numbers:
    result = aliquot_sequence(n)
    print(f"Starting number: {n}")
    print(f"Aliquot sequence: {result}")
    print(f"Number of terms: {len(result)}")
    print()
```
